```python
#Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Import dataset
df = pd.read_csv("Churn_Modelling.csv")
x = df.iloc[:,3:13].values
y = df.iloc[:,13].values
```

```python
df.head()
```

```python
#Encoding Categorical data
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
Labelencoder_x1=LabelEncoder()
x[:,1] = Labelencoder_x1.fit_transform(x[:,1])
Labelencoder_x2 = LabelEncoder()
x[:,2] = Labelencoder_x2.fit_transform(x[:,2])
x
```

```python
#Splitting the dataset into training and testing
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
len(df)
```

```python
len(x_test)
```

```python
#Feature Scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
#Classical ML Model
from sklearn.tree import DecisionTreeClassifier
clf = DecisionTreeClassifier()
clf.fit(x_train,y_train)
pred = clf.predict(x_test)
```

```python
#Working on Confusion Matrix
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, pred)
print(cm)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, pred)
```

```python
#Now build ANN
#Import keras libraries and packages
import keras
from keras.models import Sequential
from keras.layers import Dense
```

```python
#Initialize ANN
classifier = Sequential()
#units: Its an art, comes by experience, input+output/2
#initializer: how much your weights are updated
#relu activation function
#input_dim: input shape (shape of data, how many features do we have in the data)
#adding the Input layer
classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu', input_dim = 10))
#adding the First hidden layer
classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu'))
#adding the Second hidden layer
classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu'))
#Adding the output
classifier.add(Dense(units = 1, kernel_initializer = 'uniform', activation = 'sigmoid'))
```

```python
#Compiling the ANN
classifier.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
y_train[0]
```

```python
#Fitting ANN for Training Set
classifier.fit(x_train, y_train, batch_size = 10, epochs = 50)
```

```python
#Making Prediciton & Evaluating model
y_pred_ann = classifier.predict(x_test)
y_pred_ann = (y_pred_ann > 0.5)

cm3 = confusion_matrix(y_test, y_pred_ann)
print(cm3)
```

```python
#Accuracy Score
from sklearn.metrics import accuracy_score
score_3 = accuracy_score(y_test, y_pred_ann)
print(score_3)
```

#HYPERPARAMETER Tuning - Run at your own risk

```python
#Importing some libraries for k-folds
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import cross_val_score
from keras.models import Sequential
from keras.layers import Dense

def build_classifier():
    classifier = Sequential()
    classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu', input_dim = 10))
    classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu'))
    classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu'))
    classifier.add(Dense(units = 1, kernel_initializer = 'uniform', activation = 'sigmoid'))
    classifier.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
    return classifier
classifier = KerasClassifier(build_fn = build_classifier, batch_size = 10, epochs = 10)
accuracies = cross_val_score(estimator = classifier, X = x_train, y = y_train, cv = 10, n_jobs = -1)
mean1 = accuracies.mean()
variance = accuracies.std()
```

```python
#Imporvising the ANN -hyper parameter optimization

from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV
from keras.models import Sequential
from keras.layers import Dense

def build_classifier():
    classifier = Sequential()
    classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu', input_dim = 10))
    classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu'))
    classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu'))
    classifier.add(Dense(units = 1, kernel_initializer = 'uniform', activation = 'sigmoid'))
    classifier.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
    return classifier
classifier = KerasClassifier(build_fn = build_classifier, batch_size = 10, epochs = 10)

parameters = {'batch_size' : [10,20],'epochs': [5,6]}

#Total Combinations = 4

grid_search =GridSearchCV(estimator = classifier, param_grid = parameters, scoring ='accuracy', n_jobs =-1, cv =5)
grid_search =grid_search.fit(x_train,y_train)
grid_search.best_params_
grid_search.best_score_
              
```
